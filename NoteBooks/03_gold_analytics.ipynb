```python
%sql
CREATE OR REPLACE TABLE workspace.gold.daily_sales
USING DELTA
AS
SELECT
CAST(order_date AS DATE) AS order_day,
COUNT(DISTINCT order_id) AS total_orders,
SUM(total_amount) AS total_revenue,
COUNT(DISTINCT customer_id) AS unique_customers
FROM workspace.silver.orders
WHERE status = 'delivered'
GROUP BY CAST(order_date AS DATE);
```

```python
%sql
SELECT * from workspace.gold.daily_sales LIMIT 10;
```

```python
%sql

CREATE OR REPLACE TABLE workspace.gold.product_performance
USING DELTA
AS
SELECT
    p.product_id,
    p.product_name,
    p.category,
    SUM(oi.quantity) AS units_sold,
    SUM(oi.line_total) AS total_sales
FROM workspace.silver.order_items oi
JOIN workspace.silver.orders o
    ON oi.order_id = o.order_id
JOIN workspace.silver.products p
    ON oi.product_id = p.product_id
WHERE o.status = 'delivered'
GROUP BY
    p.product_id,
    p.product_name,
    p.category;
```

```python
%sql

CREATE OR REPLACE TABLE workspace.gold.country_revenue
USING DELTA
AS
SELECT
    c.country,
    COUNT(DISTINCT o.order_id) AS total_orders,
    SUM(o.total_amount) AS total_revenue
FROM workspace.silver.orders o
JOIN workspace.silver.customers c
    ON o.customer_id = c.customer_id
WHERE o.status = 'delivered'
GROUP BY c.country;
```
